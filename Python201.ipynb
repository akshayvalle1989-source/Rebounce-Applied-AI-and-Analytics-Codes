Text Article Analyzer
A Python program that reads a text file and generates a basic text analysis report.

The program analyzes the contents of a text file and provides information such as:

Total number of words
Total number of sentences
Average word length
Estimated reading time
Most frequently used words


```python
import string
a=input("enter a paragraph:")
b=a.lower().split()
print("word count:",len(b))
print("sentence count:",len(a.split(",")))
b=[c.strip(string.punctuation) for c in b]
word_counts= {}
for c in b:
 if c in word_counts:
    word_counts[c]+=1
 else:
    word_counts[c]=1
sorted_words=sorted(word_counts.items(),key=lambda x:x[1],reverse=True)
top_10=sorted_words[:10]
print("top 10 words:",top_10)
average_word_length=sum(len(word) for word in b)/len(b)
print("average word length:",average_word_length)

# Reading time estimate (200 words per minute)
reading_speed = 200
reading_time = len(b) / reading_speed
print("estimated reading time (minutes):", round(reading_time, 2))
```
