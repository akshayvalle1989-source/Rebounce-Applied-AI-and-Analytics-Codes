```python
import pandas as pd
import numpy as np
```

```python
url = "https://huggingface.co/datasets/aarav912/online-retail/resolve/main/online_retail.csv"
df = pd.read_csv(url)
```

# **6. Create Sales Column**

```python
df['Sales'] = df['Quantity'] * df['UnitPrice']
print(df.shape)
df.head()
```

# **7. NumPy Analysis**

```python
sales = df['Sales'].values
mean =round(np.mean(sales),2)
median = round(np.median(sales),2)
minimum = round(np.min(sales),2)
maximum = round(np.max(sales),2)
std_dev = round(np.std(sales),2)
print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
print(f"Standard Deviation: {std_dev}")
```

Mean > Median:
The data is right-skewed (positively skewed)
There are outliers on the high end pulling the mean upward,although most sales are lower,few high sales are present

# **9. GroupBy & Aggregation**

```python
country_summary = df.groupby('Country').agg(
    Number_of_Transactions=('Sales', 'count'),
    Total_Sales=('Sales', 'sum'),
    Average_Sales=('Sales', 'mean')
).reset_index()
print(country_summary)
```

```python
country_summary = country_summary.sort_values('Total_Sales', ascending=False)
print(country_summary.head())
```

```python
country_summary = country_summary.sort_values('Average_Sales', ascending=False)
print(country_summary.head())
```

```python
country_summary = country_summary.sort_values('Total_Sales', ascending=False)
print(country_summary.head(1))
```

```python
region_lookup = {
"United Kingdom": "Europe",
"France": "Europe",
"Germany": "Europe",
"Spain": "Europe",
"Netherlands": "Europe",
"Belgium": "Europe",
"Switzerland": "Europe",
"Portugal": "Europe",
"Australia": "Oceania",
"Japan": "Asia",
"USA": "Americas",
}
region_df = pd.DataFrame(list(region_lookup.items()), columns=['Country', 'Region'])

print(region_df)
```

# **10. Merge / Region Analysis**

```python
merged= pd.merge(country_summary, region_df, on="Country", how="left")
merged
```

```python
merged.fillna("Other")
```

```python
report=merged.groupby("Region").agg(
    Total_Sales=("Total_Sales", "sum"),
 ).reset_index()
print(report)
```

```python
sorted_report=report.sort_values("Total_Sales", ascending=False)
print(sorted_report)
```
